```python
import sqlite3

conn = sqlite3.connect("sales_data.db")
cursor = conn.cursor()

# Products table
cursor.execute("""
CREATE TABLE IF NOT EXISTS products (
    product_id INTEGER PRIMARY KEY,
    name TEXT,
    price REAL,
    stock INTEGER
)
""")

# Orders table
cursor.execute("""
CREATE TABLE IF NOT EXISTS orders (
    order_id INTEGER PRIMARY KEY,
    product_id INTEGER,
    quantity INTEGER,
    customer_id INTEGER,
    date TEXT
)
""")

products = [
    (1, "Laptop", 800, 20),
    (2, "Headphones", 150, 50),
    (3, "Keyboard", 70, 40)
]

orders = [
    (1, 1, 2, 101, "2025-01-10"),
    (2, 2, 1, 102, "2025-01-12"),
    (3, 1, 3, 103, "2025-01-15")
]

cursor.executemany("INSERT OR IGNORE INTO products VALUES (?,?,?,?)", products)
cursor.executemany("INSERT OR IGNORE INTO orders VALUES (?,?,?,?,?)", orders)

conn.commit()
conn.close()

print("Database created successfully")
```

```python
db = SQLDatabase.from_uri("sqlite:///sales_data.db")
print(db.get_table_info())
```

```python
db_chain = SQLDatabaseChain.from_llm(
    llm,
    db,
    verbose=True,
    return_direct=True
)
```

```python
def query_database(question):
    result = db_chain.run(question)
    return f"Database result: {result}"
```

```python
from ddgs import DDGS

def internet_search(query: str):
    with DDGS() as ddgs:
        results = list(ddgs.text(query, max_results=5))

    output = []
    for r in results:
        text = f"""
Title: {r.get('title')}
Snippet: {r.get('body')}
Source: {r.get('href')}
"""
        output.append(text)

    return "\n".join(output)
```

```python
internet_search("What is the capital of France?")
```

```python
internet_search("capital of India")
```

```python
def query_database(question: str):
    result = db_chain.run(question)
    return str(result)
```

```python
database_tool = Tool(
    name="database_query",
    func=query_database,
    description="""
Use this tool for questions about company products, stock levels, prices, or orders.
The input must be the full user question.
Example inputs:
- How many units of Laptop are in stock?
- What is the price of Headphones?
"""
)
```

```python
internet_tool = Tool(
    name="internet_search",
    func=internet_search,
    description="Use this tool for answering general knowledge or current events questions."
)
tools = [internet_tool, database_tool]
```

```python
from langchain_core.prompts import PromptTemplate

template = """
You are an intelligent AI assistant that answers user questions using tools when needed.

You have access to the following tools:

{tools}

You can use the following tool names:
{tool_names}

Follow this reasoning format:

Question: the user question
Thought: think about what to do
Action: choose one of [{tool_names}]
Action Input: the input to the action
Observation: the result of the action

Repeat the Thought/Action/Action Input/Observation steps if needed.

When you know the final answer respond with:

Thought: I now know the final answer
Final Answer: the answer to the user

Begin!

Question: {input}
Thought: {agent_scratchpad}
"""

prompt = PromptTemplate.from_template(template)
```

```python
from langchain.agents import create_react_agent

agent = create_react_agent(
    llm=llm,
    tools=tools,
    prompt=prompt
)
```

```python
from langchain.agents import AgentExecutor

agent_executor = AgentExecutor(
    agent=agent,
    tools=tools,
    verbose=True,
    max_iterations=4,
    handle_parsing_errors=True
)
```

```python
agent_executor.invoke(
    {"input": "How many units of laptop are in stock?"}
)
```

```python
agent_executor.invoke(
    {"input": "What is the capital of france?"}
)
```
